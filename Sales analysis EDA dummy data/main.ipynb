```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
sales_data = {
    "Month": [
        "January",
        "February",
        "March",
        "April",
        "May",
        "June"
    ],

    "Laptop": [15, 18, 20, 22, 25, 28],

    "Mobile": [25, 28, 30, 32, 35, 38],

    "Tablet": [10, 12, 15, 18, 20, 22],

    "Headphones": [20, 22, 25, 27, 30, 32],

    "Smartwatch": [12, 15, 18, 20, 23, 26]
}

df = pd.DataFrame(sales_data)

df
```

```python
print(df.Laptop.sum())
print(df.Laptop.mean())
print(df.Laptop.max())
```

```python
'''
Laptop = Rs. 80,000
Mobile = Rs. 40,000
Tablet = Rs. 30,000
Headphones = Rs. 10,000
Smartwatch = Rs. 15,000

'''
df["LaptopRevenue"] = df["Laptop"] * 80000
df["MobileRevenue"] = df["Mobile"] * 40000
df["TabletRevenue"] = df["Tablet"] * 30000
df["HeadphonesRevenue"] = df["Headphones"] * 10000
df["SmartwatchRevenue"] = df["Smartwatch"] * 15000

df
```

```python
# monthly revenue
df["TotalRevenue"] = (
    df["LaptopRevenue"] +
    df["MobileRevenue"] +
    df["TabletRevenue"] +
    df["HeadphonesRevenue"] +
    df["SmartwatchRevenue"]
)

df.head()
```

```python
df.describe()
```

```python
highest_month = df.loc[df["TotalRevenue"].idxmax()]
highest_month["Month"]
```

```python
#plt.style.use('fivethirtyeight')
plt.plot(df["Month"],df["TotalRevenue"],marker="o")
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.tight_layout()
plt.grid(True)

plt.show()
```

```python
revenue = [df["LaptopRevenue"].sum(),df["MobileRevenue"].sum(),df["TabletRevenue"].sum(),df["HeadphonesRevenue"].sum(),df["SmartwatchRevenue"].sum()]

products = ["Laptop","Mobile","Tablet","Headphones","Smartwatch"]
```

```python
product_revenue = {
    "Laptop": df["LaptopRevenue"].sum(),
    "Mobile": df["MobileRevenue"].sum(),
    "Tablet": df["TabletRevenue"].sum(),
    "Headphones": df["HeadphonesRevenue"].sum(),
    "Smartwatch": df["SmartwatchRevenue"].sum()
}
```

```python
plt.figure(figsize=(8, 8))

plt.pie(
    product_revenue.values(),
    labels=product_revenue.keys(),
    autopct="%1.1f%%"
)

plt.title("Product Revenue Contribution")

plt.show()
```

```python
q1_revenue = df["TotalRevenue"].iloc[0:3].sum()
q2_revenue = df["TotalRevenue"].iloc[3:6].sum()

print("Q1 Revenue:", q1_revenue)
print("Q2 Revenue:", q2_revenue)
```

```python
quarters = ["Q1", "Q2"]
quarterly_revenue = [q1_revenue, q2_revenue]

plt.figure(figsize=(7, 5))

plt.bar(quarters, quarterly_revenue)

plt.title("Quarterly Revenue Comparison")
plt.xlabel("Quarter")
plt.ylabel("Revenue")

plt.show()
```
