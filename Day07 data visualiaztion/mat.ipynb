```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(9,5))

plt.subplot(1,2,1)
plt.plot([1,2,3,4,5],[2,4,6,8,10],color ='b',linestyle=':',marker='o',alpha=1,linewidth=3)
plt.xlabel('Cont')
plt.ylabel('Square')
plt.title('Line plot')
plt.grid()


plt.subplot(1,2,2)
plt.plot([1,2,3,4,5],[0,12,18,20,22],color ='r',linestyle='--',marker='o',alpha=1,linewidth=4,markersize=10)
plt.xlabel('Cont')
plt.ylabel('Square')
plt.title('Line plot')
plt.grid()

plt.tight_layout()



plt.show()
```

```python
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
sns.pairplot(tips)
plt.show()
```

```python
sns.histplot(tips['total_bill'],bins=10,kde=True)
```

```python
tips.head()
```

```python
sns.violinplot(y='total_bill',x='day',data=tips)
```

```python
sns.lineplot(x='tip', y='total_bill', data=tips, hue='smoker')
plt.grid()
plt.tight_layout()
```

```python
corr = tips.select_dtypes(include=['float64','int64']).corr()
```
