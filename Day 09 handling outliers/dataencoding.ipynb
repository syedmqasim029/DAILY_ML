```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd
```

```python
data = pd.DataFrame(
    {
        'color':['red','blue','green','green','red','blue']
    }
)
```

```python
data
```

```python
encoder = OneHotEncoder()
encoded_df = encoder.fit_transform(data).toarray()
```

```python
encoded_df = pd.DataFrame(encoded_df,columns=encoder.get_feature_names_out())
```

```python
encoded_df
```

```python
import seaborn as sns 
df = sns.load_dataset('tips')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
encoder = OneHotEncoder()
encodeed = encoder.fit_transform(df[['sex','smoker','day','time']]).toarray()
```

```python
encoded_df = pd.DataFrame(encodeed,columns=encoder.get_feature_names_out())
encoded_df
```

```python
full_df = pd.concat([df,encoded_df],axis=1)
```
