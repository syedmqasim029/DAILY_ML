```python
# numpy vectorized operations
ar1 = np.array([i for i in range(1,6)])
ar2 = np.array([i for i in range(6,11)])
print(ar1-ar2)
print(ar1+ar2)
print(ar1/ar2)
print(ar1%ar2)
```
